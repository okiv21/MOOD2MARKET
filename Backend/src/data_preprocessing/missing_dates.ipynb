```python
import pandas as pd
```

```python
start_date = "2024-12-26"
end_date = "2026-03-05"

full_dates = pd.date_range(start=start_date, end=end_date)
```

```python
#explicitly mark the missing dates
reddit_df = pd.read_csv("C:\\Users\\OKIO\\Desktop\\mood2market\\Backend\\Data\\processed\\daily_reddit_sentiment.csv")
news_df = pd.read_csv("C:\\Users\\OKIO\\Desktop\\mood2market\\Backend\\Data\\processed\\daily_news_sentiment.csv")

#convert date columns to datetime
reddit_df['date'] = pd.to_datetime(reddit_df['date'])
news_df['date'] = pd.to_datetime(news_df['date'])

# Aggregate daily averages for news and reddit
news_daily = news_df.groupby('date').agg(
    {'sentiment_score': 'mean',
      'sentiment_positive': 'mean',
      'sentiment_negative': 'mean',
      'sentiment_neutral': 'mean', 
      'news_count': 'sum'})

reddit_daily = reddit_df.groupby('date').agg(
    {'sentiment_score': 'mean',
      'sentiment_positive': 'mean',
      'sentiment_negative': 'mean',
      'sentiment_neutral': 'mean', 
      'reddit_count': 'sum'})

#Create full daily date range
full_range = pd.date_range(start= min(news_daily.index.min(), reddit_daily.index.min())
                          , end= max(news_daily.index.max(), reddit_daily.index.max()), freq="D")
#Reindex to full date range, filling missing dates with NaN
reddit_daily = reddit_daily.reindex(full_range)
reddit_daily['reddit_count'] = reddit_daily['reddit_count'].fillna(0)
#news: forward filling
news_daily = news_daily.reindex(full_range).ffill()

news_daily = news_daily.reset_index().rename(columns={'index':'date'})
reddit_daily = reddit_daily.reset_index().rename(columns={'index':'date'})



```

```python
reddit_daily.to_csv(r"C:\Users\OKIO\Desktop\mood2market\Backend\Data\processed\main Data\reddit_marked.csv", index=False)
news_daily.to_csv(r"C:\Users\OKIO\Desktop\mood2market\Backend\Data\processed\main Data\news_filled.csv", index=False)
print("Reddit and news sentiment data with missing dates handled have been saved successfully.")
```
